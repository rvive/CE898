```python
import numpy as np
import pandas as pd

# Inflow hydrograph
time = np.arange(0, 311, 10)

I = np.array([10, 15, 24, 38, 50, 78, 70, 45,
              35, 25, 20, 15, 13, 12, 11, 10,
              10, 10, 10, 10, 10, 10, 10, 10,
              10, 10, 10, 10, 10, 10, 10, 10],dtype=float)

# Reservoir data
Q_curve = np.array(
    [0, 3, 8, 17, 30, 43, 60, 78, 97, 117, 137],dtype=float)

S_curve = np.array([0, 43500, 87120, 130680, 174240,
                    217800, 261360, 304920, 348480,
                    392040, 435600], dtype=float)

# Time step: 10 min = 600 s
dt = 600

# Storage-indication function
F_curve = 2 * S_curve / dt + Q_curve

# Arrays for results
Q = np.zeros(len(time))
I_sum = np.full(len(time), np.nan)
term_previous = np.full(len(time), np.nan)
term_current = np.full(len(time), np.nan)

# Initial condition
Q[0] = 10
# Initial storage corresponding to Q = 10
S0 = np.interp(Q[0], Q_curve, S_curve)
# Initial value: 2S1/dt - Q1
previous = 2 * S0 / dt - Q[0]

# Routing calculations
for i in range(1, len(time)):
    # I1 + I2
    I_sum[i] = I[i-1] + I[i]
    # 2S1/dt - Q1
    term_previous[i] = previous
    # 2S2/dt + Q2
    term_current[i] = I_sum[i] + term_previous[i]
    # Obtain Q2 from the storage-indication relationship
    Q[i] = np.interp(term_current[i], F_curve, Q_curve)
    previous = term_current[i] - 2 * Q[i]

results = pd.DataFrame({
    "Time (min)": time,
    "Inflow (m3/s)": I,
    "I1 + I2": I_sum,
    "2S1/dt - Q1": term_previous,
    "2S2/dt + Q2": term_current,
    "Outflow (m3/s)": Q})

print(results.round(3).to_string(index=False))
```

```python
import numpy as np

time = np.arange(1, 25)

I = np.array([
    75, 80, 100, 130, 170, 305,
    465, 405, 355, 280, 205, 160,
    120, 75, 75, 75, 75, 75,
    75, 75, 75, 75, 75, 75
], dtype=float)

C0 = -0.0526316
C1 =  0.5789474
C2 =  0.4736842

Q = np.zeros(len(I))
Q[0] = I[0]

for i in range(1, len(I)):

    Q[i] = (
        C0 * I[i]
        + C1 * I[i-1]
        + C2 * Q[i-1]
    )

for t, inflow, outflow in zip(time, I, Q):
    print(f"{t:2.0f}  {inflow:7.1f}  {outflow:8.3f}")
```
